```python
nums = [1,2,3,1,1,3,4,5,6]
val = 1
def removeElement( nums, val):
    pos = 0
    for i in range(len(nums)):
        if nums[i] != val:
            nums[pos] , nums[i] = nums[i] , nums[pos]
            pos +=1 
    return pos
    

k = removeElement(nums , 1)
print(k)
```
