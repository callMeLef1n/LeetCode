```python
def merge(nums1, m, nums2, n):
    l1 = m - 1
    l2 = n - 1
    k = m + n - 1

    while l2 >= 0:
        if l1 >= 0 and nums1[l1] >= nums2[l2]:
            nums1[k] = nums1[l1]
            l1 -= 1
        else:
            nums1[k] = nums2[l2]
            l2 -= 1
        k -= 1  
        
    return nums1

 
```
