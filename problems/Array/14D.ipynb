```python
strs = ["dog","racecar","car"]
def longestCommonPrefix(strs):
    final = ""
    for i in range(len(strs[0])):
        final += strs[0][i]
        for word in strs :
            if word.startswith(final):
                continue
            else:
                final = final[:-1]
    return final

longestCommonPrefix(strs)
```
