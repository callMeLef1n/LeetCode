```python

def searchInsert(self, nums, target):
    left = 0 
    right = len(nums) - 1
    mid = (left + right) // 2
    while left <= right:
        if target == nums[mid]:
            return mid
        if nums[mid] > target:
            right = mid - 1
            mid = (right+left)//2
        if nums[mid] < target:
                left = mid +1
                mid = (right+left)//2
    return left
        
        
```
