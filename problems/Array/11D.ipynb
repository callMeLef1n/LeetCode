```python
height = [1,1]
def maxArea(height):
        max = 0
        for i in range(len(height)):
            for j in range(len(height)):
                if i == j:
                     continue
                if height[i] > height[j]:
                    h = height[j]*abs(j-i)
                    if h > max:
                         max = h
                if height[i] <= height[j]:
                    h = height[i]*abs(i-j)
                    if h > max:
                        max = h
        return max
maxArea(height)
```

```python
def maxArea(height):
        left = 0
        right = len(height) - 1
        max_water = 0

        while left < right:
            h = min(height[left], height[right])
            width = right - left

            water = h * width

            if water > max_water:
                max_water = water

            if height[left] < height[right]:
                left += 1
            else:
                right -= 1

        return max_water
```
