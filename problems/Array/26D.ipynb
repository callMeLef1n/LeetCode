```python
def removeDuplicates(nums):
    left = 0
    right = left + 1
    num = len(nums)
    while right < num:
        if nums[left] == nums[right]:
            for i in range(left , num -1):
                nums[i] = nums[i+1]
            num-= 1
        else:
            left +=1
            right+=1
    return num
```

```python
def removeDuplicates(nums):
    left = 0
    right = left + 1
    if len(nums) == 0:
        return 0
    num = 1
    while right < len(nums):
        if nums[left] == nums[right]:
            right += 1
        else:
            nums[left+1] = nums[right]
            num +=1
            left +=1
            right += 1
    return num
```
